## Basic Commands

 1 - Docker Pull Command
* ``` docker pull docker/getting-started ```

 2 - Check Docker version
* ``` docker -v ```

 3 - Run Docker image (-d = deteched mode, -p = assigning port 90:90 (first 90 for host port and second 90 for container port))
* ``` docker run -d -p 80:80 (image name here) ```
* ``` docker run -d -p 80:80 docker/getting-started ```

 4 - Checking downloaded images
* ``` docker images ```

 5 - Checking running containers
* ``` docker ps ```

6 - Stop container
* ``` docker stop (container id here) ```

7 - Remove docker image
* ``` docker rm (image id here) ```

8 - Remove docker image forcefully
* ``` docker rm -f (image id here) ```

9 - Run without pulling (it will pull if not exist and then  run it)
* ` docker run hello-world `
